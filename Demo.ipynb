# Поиск похожих и дублирующихся фотографий

Запуск проекта через командную строку. Ячейки выполнять по порядку.

## Подготовка

```python
import os
import subprocess

DOTNET = "dotnet"
PHOTOS = os.path.abspath("DemoPhotos")

CONSOLE_APP = os.path.abspath(os.path.join("PhotoDuplicates.ConsoleApp", "bin", "Release", "net10.0-windows", "PhotoDuplicates.ConsoleApp.exe"))
GUI_APP = os.path.abspath(os.path.join("PhotoDuplicates.WinForms", "bin", "Release", "net10.0-windows", "PhotoDuplicates.exe"))

def run(command):
    print(">", " ".join(command))
    result = subprocess.run(command, capture_output=True, text=True, encoding="utf-8", errors="replace")
    print(result.stdout)
    if result.stderr:
        print(result.stderr)
    print("Код возврата:", result.returncode)

def show(*args):
    command = [GUI_APP] + list(args)
    print(">", " ".join(command))
    subprocess.Popen(command)

print("Папка с фото:", PHOTOS)
```

## Сборка и тесты

```python
run([DOTNET, "build", "PhotoDuplicates.slnx", "-c", "Release", "--nologo", "-v", "quiet"])
```

```python
run([DOTNET, "test", "PhotoDuplicates.slnx", "-c", "Release", "--nologo", "--no-build", "--logger", "console;verbosity=detailed"])
```

## Консольный режим

Справка:

```python
run([CONSOLE_APP, "--help"])
```

Поиск с очисткой кеша, все хеши считаются заново:

```python
run([CONSOLE_APP, PHOTOS, "--clear-cache"])
```

Повторный поиск. Теперь хеши берутся из кеша, это видно в строке "из кеша":

```python
run([CONSOLE_APP, PHOTOS])
```

Только точные копии. Картинки не открываются, считается только SHA-256:

```python
run([CONSOLE_APP, PHOTOS, "--exact-only"])
```

Порог 0, похожими считаются только картинки с одинаковым dHash:

```python
run([CONSOLE_APP, PHOTOS, "--threshold", "0"])
```

Без вложенных папок, отчёт сохраняется в файл:

```python
run([CONSOLE_APP, PHOTOS, "--no-recursive", "--report", "report.txt"])
print(open("report.txt", encoding="utf-8-sig").read()[:600])
```

Ошибки в параметрах:

```python
run([CONSOLE_APP, PHOTOS, "--threshold", "100", "--fast"])
```

```python
run([CONSOLE_APP, r"C:\нет\такой\папки"])
```

## Окно программы

Окно сразу начинает поиск в указанной папке. F6 отмечает дубликаты, Del удаляет отмеченные в корзину, Ctrl+M перемещает, F1 справка.

```python
show(PHOTOS)
```

Только точные копии, без вложенных папок:

```python
show(PHOTOS, "--exact-only", "--no-recursive")
```

Справка:

```python
show("--help")
```
